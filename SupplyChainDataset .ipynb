## 2️⃣ Data Cleaning & Feature Engineering
*Removing redundant attributes and PII columns, standardizing schema names to `snake_case`, parsing datetime timestamps, and engineering shipping SLA delay metrics (`shipping_delay_days`).*

```python
(df['Order Item Product Price'] != df['Product Price']).sum()
```

```python
(df['Order Customer Id'] != df['Customer Id']).sum()
(df['Sales per customer'] != df['Order Item Total']).sum()
```

```python
df_clean = df.drop(columns=['Order Item Product Price' , 'Product Description' , 'Customer Email' , 'Order Zipcode' , 'Product Image' , 'Customer Password' , 'Product Status' , 'Customer Street','Customer Zipcode', 'Order Profit Per Order' , 'Product Card Id', 'Order Customer Id' ,'Order Item Total' ])
```

```python
df_clean.shape
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip().str.lower().str.replace(' ', '_').str.replace('(', '', regex=False).str.replace(')', '', regex=False))
```

```python
df_clean.columns
```

```python
df_clean.drop(columns=["product_category_id"],inplace=True)
```

```python
df_clean["customer_full_name"] =( df_clean['customer_fname'].fillna('')+ " " + df_clean['customer_lname'].fillna('')).str.strip()
```

```python
df_clean["customer_full_name"].isnull().sum()
```

```python
df_clean.drop(columns=['customer_fname','customer_lname'],inplace=True)
```

```python
df_clean.info()
```

```python
df_clean.rename(columns={
    'order_date_dateorders': 'order_date','shipping_date_dateorders': 'shipping_date'}, inplace=True)

df_clean['order_date'] = pd.to_datetime(df_clean['order_date'])
df_clean['shipping_date'] = pd.to_datetime(df_clean['shipping_date'])


df_clean['shipping_delay_days'] = (
    df_clean['days_for_shipping_real'] - df_clean['days_for_shipment_scheduled'])
```

# 📊 Key Insights & Business Recommendations

### 1. Data Preparation & Feature Engineering Summary
* **Dataset Scale:** Cleaned and standardized **180,519** supply chain records from 53 raw columns down to **41** high-value analytical features with **0 missing values**.
* **Privacy & Schema Hygiene:** Removed redundant attributes and sensitive PII fields (`Customer Email`, `Customer Password`, `Customer Street`, `Customer Zipcode`) and standardized all column names to `snake_case`.
* **Engineered Features:**
  * `customer_full_name`: Consolidated customer first and last names cleanly.
  * `shipping_delay_days`: Calculated actual vs. scheduled shipping variance (`days_for_shipping_real - days_for_shipment_scheduled`) to evaluate SLA compliance.
  * `order_year` & `order_month`: Extracted temporal dimensions from `datetime64` timestamps for seasonality and trend tracking.

---

### 2. Key Analytical Insights (EDA Findings)
* **Critical SLA & Delivery Bottlenecks:**
  * **54.83%** of all orders (**98,977** orders) experience delivery delays (`Late delivery` is the dominant status across the dataset).
  * **First Class** (**95.32%** late rate, avg **+1.00** day delay) and **Second Class** (**76.63%** late rate, avg **+1.99** days delay) suffer from significantly higher SLA breaches compared to **Standard Class** (**38.07%**), pointing to unrealistic scheduled SLA targets (`days_for_shipment_scheduled`) rather than purely carrier slowness.
* **Profitability & Discount Dynamics:**
  * **18.71%** of all orders (**33,784** orders) are fulfilled at a net financial loss (`benefit_per_order < 0`).
  * Correlation between discount rate and order profit is negligible (**-0.018**), proving that heavy discounting is **not** the primary driver of negative margins.
  * Losses are concentrated in specific high-cost/low-margin SKUs — only **3 products** generate a net negative cumulative profit:
    1. `SOLE E35 Elliptical` (**-965.12**)
    2. `Bushnell Pro X7 Jolt Slope Rangefinder` (**-255.95**)
    3. `SOLE E25 Elliptical` (**-169.56**)
* **Revenue & Category Concentration:**
  * **Fan Shop** (~1.83M) and **Apparel** (~881.9K) lead overall department profitability.
  * **Fishing** (~6.93M), **Cleats** (~4.43M), and **Camping & Hiking** (~4.12M) are the top revenue-generating product categories.

---

### 3. Actionable Business Recommendations
1. **Recalibrate Shipping SLAs:** Adjust scheduled delivery windows (`days_for_shipment_scheduled`) for *First Class* and *Second Class* tiers to match realistic carrier transit times and reduce SLA breaches.
2. **Audit Loss-Making SKUs:** Review pricing, supplier costs, and shipping fees for `SOLE E35 Elliptical`, `SOLE E25 Elliptical`, and `Bushnell Pro X7` to eliminate structural per-unit losses.
3. **Protect High-Margin Core Departments:** Prioritize inventory availability and fulfillment speed for **Fan Shop** and **Fishing** categories, as they drive the bulk of company revenue and net profit.
